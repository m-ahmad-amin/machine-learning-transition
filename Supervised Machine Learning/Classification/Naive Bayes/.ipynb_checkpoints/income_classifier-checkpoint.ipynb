```python
X_train = train_df.drop('income', axis=1)
y_train = train_df['income']
```

```python
test_df = pd.read_csv('data/adult.test', header=None, skiprows=1)
```

```python
test_df.head()
```

```python
test_df.columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]
```

```python
test_df.head()
```

```python
test_df['income'] = test_df['income'].str.replace('.', '', regex=False)
```

```python
test_df.head()
```

```python
X_test = test_df.drop('income', axis=1)
y_test = test_df['income']
```

```python
numerical_features = [
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]

categorical_features = [
    "workclass",
    "education",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "native_country"
]
```

```python
X_train_num = X_train[numerical_features]
X_train_cat = X_train[categorical_features]

X_test_num = X_test[numerical_features]
X_test_cat = X_test[categorical_features]
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Encode categorical features
encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_train_cat = encoder.fit_transform(X_train_cat)
X_test_cat = encoder.transform(X_test_cat)

# Shift values so unknown (-1) becomes 0
# and known categories start from 1
X_train_cat = X_train_cat.astype(int) + 1
X_test_cat = X_test_cat.astype(int) + 1

# Check
print(X_train_cat[:5])
print(X_test_cat[:5])
print(encoder.categories_)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)
```

```python
from sklearn.naive_bayes import GaussianNB, CategoricalNB

gaussian_nb = GaussianNB()
categorical_nb = CategoricalNB()
```

```python
gaussian_nb.fit(X_train_num, y_train_encoded)
categorical_nb.fit(X_train_cat, y_train_encoded)
```

```python
l1 = gaussian_nb.predict_log_proba(X_test_num)
```

```python
l2 = categorical_nb.predict_log_proba(X_test_cat)
```

```python
cp = categorical_nb.class_log_prior_
```

```python
mixed = l1 + l2 - cp
```
