```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Example data
X = np.array([
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4],
    [2000, 4],
    [2200, 5],
    [2500, 5],
    [2800, 6]
])

y = np.array([200, 250, 300, 350, 400, 450, 500, 550])

# Train model
model = LinearRegression()
model.fit(X, y)

# Create 3D plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

# Actual data points
ax.scatter(
    X[:, 0],
    X[:, 1],
    y,
    label="Actual data"
)

# Create grid for regression plane
x1 = np.linspace(X[:, 0].min(), X[:, 0].max(), 20)
x2 = np.linspace(X[:, 1].min(), X[:, 1].max(), 20)

X1, X2 = np.meshgrid(x1, x2)

# Calculate predicted y for every point on the grid
Y_pred = (
    model.intercept_
    + model.coef_[0] * X1
    + model.coef_[1] * X2
)

# Regression plane
ax.plot_surface(
    X1,
    X2,
    Y_pred,
    alpha=0.5
)

# Labels
ax.set_xlabel("Size")
ax.set_ylabel("Bedrooms")
ax.set_zlabel("Price")

plt.show()
```
