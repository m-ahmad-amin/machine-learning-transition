```python
import pandas as pd

df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore', 'Peshawar', 'Karachi'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Male'],
    'Age': [22, 25, 21, 24, 23, 26],
    'Salary': [50000, 65000, 45000, 55000, 48000, 70000]
})

print(df)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df[['City', 'Gender']])

print(encoded)
print(encoder.get_feature_names_out())
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df[['City', 'Gender']])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['City', 'Gender']),
    index=df.index
)

print(encoded_df)
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'City': ['Lahore', 'Karachi', 'Islamabad', 'Lahore', 'Peshawar', 'Karachi'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Male'],
    'Age': [22, 25, 21, 24, 23, 26],
    'Salary': [50000, 65000, 45000, 55000, 48000, 70000]
})
```

```python
df
```

```python
cat_cols = ['City', 'Gender']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            OneHotEncoder(
                drop='first',
                sparse_output=False
            ),
            cat_cols
        )
    ],
    remainder='passthrough'
).set_output(transform='pandas')
```

```python
encoded_df = preprocessor.fit_transform(df)
```

```python
encoded_df
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

train = pd.DataFrame({
    'sex': ['male', 'female'],
    'city': ['Lahore', 'Karachi']
})

ohe = OneHotEncoder(sparse_output=False)

# Fit in this order: sex, city
ohe.fit(train[['sex', 'city']])

print(ohe.categories_)
```

```python
t = ohe.transform(train[['sex', 'city']])
```
