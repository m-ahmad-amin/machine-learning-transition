```python
import pandas as pd

df = pd.DataFrame({
    'age': [22, 35, 28, 41, 30, 26, 38, 45],
    'education': [
        'High School', 'Bachelor', 'Master', 'PhD',
        'Bachelor', 'High School', 'Master', 'PhD'
    ],
    'employment_type': [
        'Private', 'Government', 'Private', 'Self-employed',
        'Government', 'Private', 'Self-employed', 'Private'
    ],
    'satisfaction': [
        'Poor', 'Good', 'Excellent', 'Average',
        'Good', 'Poor', 'Excellent', 'Average'
    ],
    'remote_worker': [
        'Yes', 'No', 'Yes', 'Yes',
        'No', 'No', 'Yes', 'No'
    ],
    'city': [
        'Lahore', 'Islamabad', 'Karachi', 'Lahore',
        'Peshawar', 'Karachi', 'Islamabad', 'Lahore'
    ],
    'salary': [
        45000, 85000, 120000, 150000,
        90000, 50000, 130000, 160000
    ]
})
```

```python
df
```

```python
df['remote_worker'].value_counts()
```

```python
# Binary categorical feature
```

```python
df['remote_worker'] = df['remote_worker'].map({
    'Yes': 1,
    'No': 0
})
```

```python
df
```

```python
df['satisfaction'].value_counts()
```

```python
df['education'].value_counts()
```

```python
# meaningful 0rder
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[
    ['High School', 'Bachelor', 'Master', 'PhD'], # school < bachelor < master < PhD
    ['Poor', 'Average', 'Good', 'Excellent'] # poor < average < good < excellent
])
```

```python
df[['education', 'satisfaction']] = oe.fit_transform(df[['education', 'satisfaction']])
```

```python
df
```

```python
df['employment_type'].value_counts()
```

```python
df['city'].value_counts()
```

```python
# nominal features -> one-hot encoding
```

```python
df = pd.get_dummies(df, columns=['employment_type', 'city'], drop_first=True, dtype=int)
```
